```python
from bs4 import BeautifulSoup
import pandas as pd
import requests

# 1. Tentukan URL target (biasanya website latihan web scraping)
url = 'https://quotes.toscrape.com/'

# 2. Ambil halaman web menggunakan requests
response = requests.get(url)

# 3. Parsing HTML dengan BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# 4. Cari elemen data yang diinginkan (misal: kutipan teks dan penulisnya)
quotes_data = []
for item in soup.find_all('div', class_='quote'):
  text = item.find('span', class_='text').get_text()
  author = item.find('small', class_='author').get_text()
  quotes_data.append({'Quote': text, 'Author': author})

# 5. Ubah ke Pandas DataFrame dan simpan ke folder data/raw
df = pd.DataFrame(quotes_data)
print(df.head())

# Menyimpan ke folder data/raw yang sudah dibuat sebelumnya
# df.to_csv('/content/drive/My Drive/proyek_akuisisi_data/data/raw/quotes.csv', index=False)
```
